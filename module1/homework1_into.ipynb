```python
med = df['horsepower'].median()
mod = df['horsepower'].mode()[0]

#df['horsepower'].fillna(mod) # google saying this returns series, technically multiple values can return if there is tie, so get 1st value only

med2 = df['horsepower'].fillna(mod).median()

print(f"Original Median: {med}")
print(f"New Median: {med2}")
print(f"Did it change? {med2 > med}")
```

```python
import numpy as np

X=df.loc[df['origin']=='Asia',['vehicle_weight','model_year']].head(7).to_numpy() #google said to_numpy better than values

XTX = np.matmul(X.T,X)

XTXinv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = np.matmul(np.matmul(XTXinv,X.T),y)

w.sum().round(4)
```
