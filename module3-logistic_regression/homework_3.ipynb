```python
import pandas as pd
import numpy as np

df = pd.read_csv("course_lead_scoring_2026.csv")
```

```python
df.head()
```

```python
# check for missing values

df.isnull().sum()
```

```python
for colname,dttype in df.dtypes.items():
    if dttype == 'str':
        df[colname] = df[colname].fillna('NA')
    elif dttype in ('float64','int64'):
        df[colname] = df[colname].fillna(0)

#more vectorized way

# Fill text/object columns
# obj_cols = df.select_dtypes(include=['object']).columns
# df[obj_cols] = df[obj_cols].fillna('NA')

# Fill numeric columns
# num_cols = df.select_dtypes(include=['number']).columns
# df[num_cols] = df[num_cols].fillna(0)
```

```python
#df['industry'].value_counts().head(1)

df['industry'].mode()
```

```python
from itertools import combinations

numer_df = df.select_dtypes(include=['number'])

numer_cols = numer_df.columns

coeff_score = []

for col1,col2 in combinations(numer_cols,2):
    coeff = df[col1].corr(df[col2])
    coeff_score.append({'col1':col1,
                  'col2':col2,
                  'coeff':coeff})
    #print(f"coefficient of {col1} and {col2} is {coeff}")

coeff_score.sort(key=lambda x:x['coeff'],reverse=True)

coeff_score
```

```python
import sklearn as skl
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42)

df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42)
```

```python
y_train = df_train.pop('converted')

y_val = df_val.pop('converted')

y_test = df_test.pop('converted')
```

```python
from sklearn.metrics import mutual_info_score

categorical = df.select_dtypes(include=['str']).columns

mutual_info = {}

for cat_col in categorical:
    mutual_info[cat_col] = mutual_info_score(y_train,df_train[cat_col])
```

```python
for key,val in dict(sorted(mutual_info.items(), key=lambda x:x[1],reverse=True)).items():
    print(f"key:{key}, value:{round(val,2)}")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing  import OneHotEncoder

coltransf = ColumnTransformer(transformers=[
                                            ('cat',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),categorical)
                                           ],
                                           remainder='passthrough')

X_train = coltransf.fit_transform(df_train)

X_train
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)

model.fit(X_train,y_train)
```

```python
y_val_pred = model.predict_proba(coltransf.fit_transform(df_val))[:,1]
```

```python
y_val_pred_final = (y_val_pred>=0.5).astype(int)
```

```python
(y_val_pred_final == y_val).astype(int).mean().round(2)
```

```python
model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)

numerical = df_train.select_dtypes(include=['number']).columns

for col in df_train.columns:
    df_train_featr = df_train.drop(columns=col)
    df_val_featr   = df_val.drop(columns=col)
    coltransf = ColumnTransformer(transformers=[
                                                ('cat',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),[item for item in categorical if item != col])
                                               ,('numer','passthrough',[item for item in numerical if item != col])
                                               ],
                                               remainder='passthrough',
                                               verbose_feature_names_out=False
                                  )
    model.fit(coltransf.fit_transform(df_train_featr),y_train)
    y_val_pred_featr = model.predict_proba(coltransf.transform(df_val_featr))[:,1]
    y_val_pred_featr_final = (y_val_pred_featr>=0.5).astype(int)
    print(f"with column/feature {col} removed, accuracy is {(y_val_pred_featr_final==y_val).astype(int).mean().round(2)}")
```

```python
for c in [0.000001, 0.00001, 0.0001, 0.001]:
    model = LogisticRegression(solver='liblinear', C=c, max_iter=1000, random_state=42)
    
    coltransf = ColumnTransformer(transformers=[
                                            ('cat',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),categorical)
                                           ],
                                           remainder='passthrough')

    X_train_reg = coltransf.fit_transform(df_train)

    model.fit(X_train_reg,y_train)

    y_val_reg = model.predict_proba(coltransf.transform(df_val))[:,1]

    y_val_reg_final = (y_val_reg>=0.5).astype(int)

    accu_c = (y_val_reg_final==y_val).astype(int).mean().round(3)

    print(f"accuracy for regularization parameter c {c} is {accu_c}")
```
